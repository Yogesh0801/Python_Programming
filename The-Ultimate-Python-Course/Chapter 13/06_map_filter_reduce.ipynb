```python
# map() means:

# Apply the same operation to every element.
#without map
l = [1, 2, 3, 4, 5]

sqList = []

for x in l:
    sqList.append(x * x)

print(sqList)
```

```python
# using map
l = [1, 2, 3, 4, 5]

def square(x):
    return x * x

sqList = map(square, l)

print(list(sqList))
```

```python
# without lambda
def square(x):
    return x * x
square(2)
```

```python
# with lambda Function
square = lambda x: x*x
print(square(5))
```

```python
add = lambda a, b: a + b

print(add(10, 20))
```

```python
# using both map and filter in list comprehension
l = [1, 2, 3, 4, 5]

square = lambda x: x*x

# print(square)

sqList = map(square, l)

print(list(sqList))
```

```python
# filter() Function
# Select only the elements that satisfy a condition.
```

```python
# Without filter()
l = [1, 2, 3, 4, 5]

even_numbers = []

for n in l:
    if n % 2 == 0:
        even_numbers.append(n)

print(even_numbers)
```

```python
# Using filter()
l = [1, 2, 3, 4, 5]
def even(n):
    if n % 2 == 0:
        return True
    return False

onlyEven = filter(even, l)

print(list(onlyEven))
# filter() asks the function for every element:

# 1 → False → Remove
# 2 → True  → Keep
# 3 → False → Remove
# 4 → True  → Keep
# 5 → False → Remove
```

```python
# filter() with Lambda

# We can make the same example shorter:

l = [1, 2, 3, 4, 5]

onlyEven = filter(lambda x: x % 2 == 0, l)

print(list(onlyEven))
```

```python
# reduce() Function
# Take multiple values and repeatedly combine them to produce one final value
```

```python
# Without reduce()
l = [1, 2, 3, 4, 5]

total = 0

for n in l:
    total = total + n

print(total)
```

```python
from functools import reduce
l = [1, 2, 3, 4, 5]
def sum(a, b):
    return a + b


print(reduce(sum, l))
```

```python
# Reduce with Multiplication

l = [1, 2, 3, 4, 5]

mul = lambda x, y: x * y

print(reduce(mul, l))
```
