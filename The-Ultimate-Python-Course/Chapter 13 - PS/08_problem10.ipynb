```python
# Add "Bhaiya" to every name

# Question: Add " Bhaiya" to every student name using map().

students = ["Ram", "Sam", "Ravi", "Amit", "John"]

result = map(lambda name: name + " Bhaiya", students)

print(list(result))
```

```python
# 2. Find students who scored more than 60

# Question: Use filter() to find marks greater than 60.

marks = [50, 60, 70, 80, 90]

result = filter(lambda mark: mark > 60, marks)

print(list(result))
```
