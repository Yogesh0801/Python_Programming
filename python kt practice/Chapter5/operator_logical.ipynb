```python
# Logical operators
# type (And, Or, Not)
# symbol (&&, ||, !)
# and (whenever both conditions are true)
# or (whenever at least one condition is true)
# not (make positive to negative and vice versa)
```

```python
# and (&&) operator
a = 20
b = 45
if a > 30 and b > 40:
    print("x")
else:
    print("y")
```

```python
# or operator
a = 20
b = 40
if a > 30 or b > 40:
    print("x")
else:
    print("y")
```

```python
# not operator
a = 40
if not a > 30:
    print("x")
```

```python
# Membership operators
# type (in, not in)
# in (if the value is present in the sequence)
# not in (if the value is not present in the sequence)

list1 = [43,54,6,8,23,4,5,6,7,8,9]
print(4 in list1)
print(10 not in list1)
```

```python
# Identity operators (it will check whether the two values are same or not with address)
# type (is, is not)
ip = 689
a = int(input("enter a value: "))
print(ip is a)
print(id(ip))
print(id(a))
```

```python
# is not operator
ip = 256
print(id(ip))
a = 256
print(id(a))
print(ip is a)
print(ip is not a)
# untill 256, python will not create a new object for the same value. It will use the same object for the same value. But after 256, it will create a new object for the same value.
```

```python
sudha = [1,2,3,4,5]
annapurna = [1,2,3,4,5]
maithli = [1,2,3,4,5]
print(id(sudha))
print(id(annapurna))
print(id(maithli))
print(sudha is not annapurna)
print(annapurna is not maithli)
print(sudha is not maithli)
```

```python
sudha = [1,2,3,4,5]
annapurna = [1,2,3,4,5]
maithli = [1,2,3,4,5]
print(sudha == annapurna)
print(annapurna == maithli)
print(maithli == sudha)
# (==) checking the values of the two objects, if they are same or not. It will return True or False.
```

```python
# bitwise operators
# bitwise operators are used to perform bit-level operations on binary numbers. The common bitwise operators in Python are:
# & (AND)
# | (OR)
# ^ (XOR)
# ~ (NOT)
# << (Left Shift)
# >> (Right Shift)
```

```python
a = 3
# binary representation of 3 is 0000 0011
print("the binary representation of 3 is : ",bin(a))
print("the binary representation of 3 is : ",format(a,'08b'))
```

```python
#& and gives 1 only when both bits is 1, otherwise it gives 0
a = 5 # 0101
b = 3 # 0011
print(5&3) 
# 0101
# 0011 
#=0001 
#=   1
```

```python
# | (OR) gives 1 if at least one of the bits is 1
print(5|3) 
# 0101 | 
# 0011 
#=0111 # 7 00000111
# = 7
```

```python
# XOR (^) gives 1 if the bits are different, otherwise it gives 0
print(5^3) 
# 0101 ^ 
# 0011 
#=0110 
# = 6 # 00000110
```
