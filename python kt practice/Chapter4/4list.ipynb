```python
# list data type data structure
# mutable
```

```python
#empty list
my_list = []
print(my_list)
print(type(my_list))
```

```python
# llist of int
my_list = [1,2,3,4,5,6]
print("int list",my_list)
# list of mixed data type
mylist = [1,"hello",6723.78]
print("list with mixed datatype",mylist)
# nested list
n_list = ["hello",[2,4,4,"python"]]
print("nested list",n_list)
```

```python
shop = ['apple','mango','carrot','banana']
print(shop)
```

```python
# append it will add a item in our list
shop.append('rice')
print(shop)
```

```python
# extend
shop.extend(['table',"chair","pen"])
print(shop)
```

```python
shop2 = ["a","b","c","d"]
print(shop2)
```

```python
# + operator we can use for add two list
```

```python
f_list = shop+shop2
print(f_list)
```

```python
# indexing
print(f_list[0])
print(f_list[1])
print(f_list[1][0])
print(f_list[4][0])
```

```python
# negetive indexing
print(f_list)
print(f_list[-1])
print(f_list[-4])
print(f_list[-5])
```

```python
#slicing
print(f_list[0:6])
# element from beginning to 3rd
print(f_list[:3])
# element from 3rd index to end
print(f_list[3:])
# another way to print all element in the list 
print(f_list[:])
```

```python
# sort : sort list in ascending order
f_list.sort()
print(f_list)
```

```python
# change item in list
f_list[2]="grapes"
print(f_list)
```

```python
# remove a element from the list
print(f_list)
f_list.remove("a")
print(f_list)
# f_list.remove("b","c") remove can take only one argument
print(f_list)
```

```python
# del : delete an item with index
del f_list[3]
print(f_list)
```

```python
# del : deleting all the item
de_list = f_list.copy()
print(de_list)
del de_list[2:5]
print(de_list)
```

```python
# pop : remove the last item
print(f_list)
x=f_list.pop()
print(f_list)
print(x)
```

```python
print(f_list)
f_list.pop(-3)
print(f_list)
```

```python
# insert : it can insert the item in particular index
f_list.insert(2,"orange")
print(f_list)
```

```python
# index
i = f_list.index("carrot")
print(i)
```

```python
j = f_list.index('banana',0,5)
print(j)
```

```python
# count: return the number of items
f_list.count('apple')
```

```python
# extend
f_list.extend([4,4,4,7,7,7,43,43,43,"apple",'Apple'])
print(f_list)
```

```python
f_list.count('Apple')
print(f_list.count(7))
print(f_list.count('pen'))
```

```python
# reverse
f_list.reverse()
print(f_list)
```

```python
a = f_list*3
print(a)
```
