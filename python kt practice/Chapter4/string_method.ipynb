### join() it will join the word to make sentences

```python
word = ["python", "is", "a", "programming", "language"]
snts = " ".join(word)
print(snts)
```

```python
path = ["C:", "Users", "John", "Documents", "file.txt"]
full_path = "\\".join(path)
print(full_path)
```

```python
# indexing and slicing
text = "Hello how are you doing today?"
```

```python
# index() returns the index of the first occurrence of a substring in a string
print(text[0])
print(text[4])
# text[4]='e' (string is immutable, so we cannot change the value of a character in a string)
print(text[4])
print(text[-1])
```

```python
# slicing returns a substring from the original string
print(text)
```

```python
print(text[5:17])
print(text[0:-1:2])
# start:ending:step
```

```python
name = "yogesh is software technical trainer"
print(len(name))
print(name.endswith("sh"))
print(name.startswith('yo'))
print(name.capitalize())
```
