```python
#indexing and slicing
week = ('mon','tue','wed','thur','fri','sat','sun')
print(week[2])
print(week[2][2])
print(week[5][0])
print(week[:3])
print(week[3:])
print(week[-1])
print(week[-4:-1])
print(week[-4])
print(week[-4][-1])
print(week[:])
print(week[1:])
print(week[-5][-2])
# indexing start from 0 index
```

```python
# tuple method
my_tuple = ('a','p','p','l','e')
# count
print(my_tuple.count('p'))
print(my_tuple.count('l'))
```

```python
#index
print(my_tuple.index('l'))
print(my_tuple.index('e'))
# print(my_tuple.index('ds'))
```

```python
# + operator concatination

tuple1 = (1,2,3,"87.43")
tuple2 = (4,5,6,"he",8.9,32,"this")
concate_tuple = tuple1+tuple2
print(concate_tuple)

```

```python
# * operator (repeat the tuple)
mult_tuple = tuple1*3
print(mult_tuple)
```

```python
# min operator (smallest value in the tuple)
n_tup = (3,6,37,9,7,)
print(n_tup)
print(min(n_tup))
```

```python
# max (largest value in the tuple)
print(max(n_tup))
```

```python
# sum (addition of all the int value)
print(sum(n_tup))
```

```python
# sorted (it will sort in ascending order)
print(n_tup)
sort_tup = sorted(n_tup)
print(sort_tup)
print(type(sort_tup))
```

```python
print(sort_tup)
```

```python
new_tuple = tuple(sort_tup)
print(new_tuple)
print(type(new_tuple))
```
