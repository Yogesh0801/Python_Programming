```python
# regex
# re
# search()
import re
text = "my phone number is 8342982320"
result = re.search(r"\d{10}",text)
print(result.group())
```

```python
# match()
import re
text = "hello India this is python"
result = re.match(r"hello",text)
print(f"found your text {result.group()}")
```

```python
# findall()
import re
# findall() returns a list of all the matches
text = "I am a student and i have 10 apples and 20 oranges and 30 bananas"
print(re.findall(r'\d+', text))
```

```python
# replace 
# it will use sub function
# re.sub(pattern, repl, string, count=0, flags=0)
import re
text = "my phone number is 0123456789"
result = re.sub(r'\d',"*@",text)
print(result)
```

```python
# \d for digit
# \w for word
# \s for space
# find the email address in the text
import re
text = "Please send an email to john@example.com"
email = re.findall(r'[\w]+@[\w]+\.[\w]+', text)
print(email)

```

```python
# match 
# ^ : start of string
# $ : end of string
# ^.{8}$ : at least 8 characters
# check password contains at least 8 characters
import re
password = "Python@123"
if re.match(r'^.{8,}$', password):
    print("password is valid")
else:
    print("password is invalid")
```

```python
# any one alphabet should be capital
# it should contain at least 8 character
# it should contain speacial keyword !@#$%^&*
# find the password pattern is correct or not
password = input('enter your password: ')
pattern = r'^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[^a-zA-Z0-9]).{8,}$'
# ^ start of string
# $ end of string
# = match exactly
# (?=.*[A-Z]) at least one capital letter
# (?=.*[a-z]) at least one lower letter
# (?=.*\d) at least one digit
# (?=.*[^a-zA-Z0-9])) at least one special character
# .{8,} at least 8 character
if re.match(pattern,password):
    print("password matched")
else:
    print("please enter strong password")
```
