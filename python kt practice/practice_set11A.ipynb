```python
# Create a class (2-D vector) and use it to create another class representing a 3-D vector.
# complex datatype 2i+3j
class TwoDvector:
    def __init__(self,i,j):
        self.i = i
        self.j = j
    def show(self):
        print(f"two D vector is {self.i}i + {self.j}j")

class ThreeDvector(TwoDvector):
    def __init__(self,i,j,k):
        super().__init__(i,j)
        self.k = k
    def show(self):
            print(f"three D vector is {self.i}i + {self.j}j + {self.k}k")

a = TwoDvector(1,2)
a.show()
b = ThreeDvector(5,2,4)
b.show()
```

```python
class Account:
    def __init__(self):
        self._balance = 0 # internal variable protected

    @property # getter : allow reading the balance like a variable
    def balance(self):
        return self._balance
    @balance.setter # how the balance is updated
    def balance(self,amount):
        if amount < 0:
            print("error : balance can not be negetive")
        else:
            self._balance = amount
acc = Account()
acc.balance = 500
print(acc.balance)
acc.balance = -100
```
