```python
# Write a program to find the greatest of four numbers entered by the user.
a1 = int(input("enter number 1: "))
a2 = int(input("enter number 2: "))
a3 = int(input("enter number 3: "))
a4 = int(input("enter number 4: "))
print(a1,a2,a3,a4)
if(a1>a2 and a1>a3 and a1>a4):
    print("Greatest number is a1 :",a1)
elif(a2>a1 and a2>a3 and a2>a4):
    print("Greatest number is a2 :",a2)
elif(a3>a1 and a3>a2 and a3>a4):
    print("Greatest number is a3 :",a3)
elif(a4>a1 and a4>a2 and a4>a3):
    print("Greatest number is a4 :",a4)
else:
    print("wrong value you have entered")
```

```python
# Write a program to find out whether a student has passed or failed if it requires a total of
# 40% and at least 33% in each subject to pass. Assume 3 subjects and take marks as an input from the user.
s1 = int(input("enter 1st subject marks: "))
s2 = int(input("enter 2st subject marks: "))
s3 = int(input("enter 3st subject marks: "))
total_percentage = (100*(s1+s2+s3))/300
print(s1,s2,s3)
print(round(total_percentage,2))
if(total_percentage>=33 and s1>=33 and s2>=33 and s3>=33):
    print("you are passed: ",total_percentage)
else:
    print("you failed, try again next time",total_percentage)
```

```python
# A spam comment is defined as a text containing following keywords: 
# “Make a lot of money”, “buy now”, “subscribe this”, “click this”. 
# Write a program to detect these spams.
p1 = "Make a lot of money"
p2 = "buy now"
p3 = "subscribe this"
p4 = "click this"
email = input("enter your comment: ")
if((p1 in email) or (p2 in email) or (p3 in email) or (p4 in email)):
    print("This email/comment is spam")
else:
    print("this email or comment is not spam")
```

```python
# Write a program to calculate the grade of a student from his marks from the following scheme:
# 90 – 100 => Ex
# 80 – 90 => A
# 70 – 80 => B
# 60 – 70 => C
# 50 – 60 => D
# <50 => F
marks = float(input("enter your marks: "))
if(marks<=100 and marks>=90):
    print("you have got extension",marks)
elif(marks<90 and marks>=80):
    print("you got A grade")
elif(marks<80 and marks>=70):
    print("you got B grade")
elif(marks<70 and marks>=60):
    print("you got C grade")
elif(marks<60 and marks>=50):
    print("you have got D grade")
else:
    print("you are failed")
```
