```python
# Transfer Statement
# Break, Continue, Pass
```

```python
# break : it will terminate the loop
for i in range(1,10):
    if i == 5:
        break
    print(i)
```

```python
# break in the list
l = [2,54,75,86,23,65]
for i in l:
    if i == 86:
        break
    print(i)
```

```python
# continue it will skip the iteration

for i in range(1,11):
    if i == 6:
        continue
    print(i) 
```

```python
l = ["apple","banana","mango","grapes"]
for i,j in enumerate(l):
    if i == "banana":
        break
    print(i,j)
```

```python
l = ["apple","banana","mango","grapes"]
for i in l:
    if i == "mango":
        continue
    print(i)
```

```python
# pass : it will not give the error
for i in range(100):
    pass
```
