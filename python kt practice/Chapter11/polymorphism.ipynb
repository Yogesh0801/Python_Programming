```python
# method overloading :
# means having the same name method is same class, but arguments are different, inheritance is not required
```

```python
class MO:
    def myfunction(self):
        print("function with zero arguments")
    def myfunction(self,a):
        print("function with one arguments")
    def myfunction(self,a,b):
        print("function with two arguments")
# create object
m = MO()
m.myfunction(10,20)
# m.myfunction()
# note : mothod overloading is not supporting by python
```

```python
# method overriding:
# means same name method, but in different class , argument also same
# inheritance is compulsary
```

```python
class mo1:
    def myfunc(self,a):
        print("class mo1 function is called")
class mo2(mo1):
    def myfunc(self,a):
        print("class mo2 function is called")
        super().myfunc(20)
class mo3(mo2):
    def myfunc(self,a):
            print("class mo3 function is called")
            # super method call to upper class method
            super().myfunc(20)
m = mo3()
m.myfunc(10)
```

```python
class bird:
    def fly(self):
        return "bird is flying"
class penguin(bird):
    def fly(self):
            return super().fly() + " penguin cannot fly"
            
p = penguin()
p.fly()
```

```python
class calculator:
    def calculate(self,a,b):
        self.a = a
        self.b = b
        c = self.a+self.b
        print(c)
class sci_calculator(calculator):
    def calculate(self,a,b):
            self.a = a
            self.b = b
            c = self.a*self.b
            print(c)
            super().calculate(2,4)
o = sci_calculator()
o.calculate(2,4)
```

```python
# static method
class employee:
    language = "python"
    salary = 12000000

    def getinfo(self):
        print(f"the language is {self.language} and salary is {self.salary}") 
    @staticmethod # decorator
    def greet():
        print("good morning")
em = employee()
# em.language
# em.salary
em.getinfo()
em.greet()
```

```python
class A:
    def cal(self,a,b):
        self.a = a
        self.b = b
        c = self.a + self.b
        print("sum ",c)
class B(A):
    def cal(self,a,b):
        self.a = a
        self.b = b
        c = self.a - self.b
        print("subtraction ",c)
        super().cal(4,2)
class C(B):
    def cal(self,a,b):
        self.a = a
        self.b = b
        c = self.a * self.b
        print("multiplication ",c)
        super().cal(4,2)
class D(C):
    def cal(self,a,b):
        self.a = a
        self.b = b
        c = self.a / self.b
        print("devide ",c)
        super().cal(4,2)
h = D()
h.cal(4,2)
```
