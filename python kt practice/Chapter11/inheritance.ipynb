```python
#single inheritance
```

```python
class ParentClass:
    def myfunction1(self):
        print("parent class function is called")
class ChildClass(ParentClass): # child class inheriting the property of parent class
    def myfunction2(self):
        print("child function is called")
# create object
c1 = ChildClass()
c1.myfunction2()
c1.myfunction1()
```

```python
# multilevel Inheritance
class A: # parent class
    def myfunc(self):
        print("Class A function called")
class B(A):
    def myfunc2(self):
        print("class B function called")
class C(B):
    def myfunc3(self):
        print("class C function called")
# create an object 
c = C()
c.myfunc2()
c.myfunc()
c.myfunc3()
```

```python
# 3.	Question: Create a class LivingBeing with a method breathe(),
#  which prints "Living beings breathe". 
# Then create a subclass Animal that inherits LivingBeing 
# and has a method move(), which prints "Animals can move".
# Finally, create another subclass Bird that inherits Animal 
# and adds a method fly(), which prints "Birds can fly". 
# Demonstrate how methods are inherited across multiple levels.
```

```python
class LivingBeing:
    def breathe(self):
            print("Living beings breathe")
class Animal(LivingBeing):
      def move(self):
            print( "Animals can move")
class Bird(Animal):
      def fly(self):
            print("Birds can fly")
b = Bird()
b.breathe()
b.fly()
b.move()
      
```

```python
# multiple Inheritance
class A1:
    def myfunc(self):
        print("CLass a function is called")
class B1:
    def myfunc2(self):
        print("class B1 function is called")
class C(A1,B1):
    def myfunc3(self):
        print("class C1 function is called")
o = C()
o.myfunc()
```

```python
# Hierarchical Inheritance
class A1: # parent class
    def Myfunc(self):
        print("Class A1 function called")
class B1(A1): # child class
    def Myfunc2(self):
        print("class B1 function called")
class B2(A1): # child class
    def myfunc3(self):
        print("class B2 function called")
class B3(A1): # child class
    def myfunc4(self):
        print("class b3 function called")
o1 = B3()
o1.myfunc4()
o1.Myfunc()

o2 = B2()
o2.myfunc3()
o2.Myfunc()
```

```python
# Hybrid inheritance
    #       parent
#       #   intermediate
    #       child

class A1: # parent class
    def myfunc(self):
        print("function of class A1")
class B1(A1):
    def myfunc2(self):
        print("function of class B1")
class B2(A1):
    def myfunc3(self):
        print("function of class B2")
class C(B1,B2):
    def myfunc4(self):
        print("function of class C is called")
o = C()
o.myfunc()
o.myfunc2()
o.myfunc3()
o.myfunc4()
```
