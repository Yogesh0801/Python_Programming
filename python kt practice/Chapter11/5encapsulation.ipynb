```python
# achive encapsualation 
# _ protected (it can access by other class)
# __ private (it cant access)
```

```python
# __ private (it cant access by anyt other class)
```

```python
class A:
    def __init__(self,a):
        self.__a = a
    def show(self):
        print("private variable: ",self.__a)

class B(A):
    def __init__(self, a):
        super().__init__(a)
    def showH(self):
        print(self.__a)
obj = B(20)
obj.show()
# obj.showH()
```

```python
# _ protected (it can access by other class)
```

```python
class A:
    def __init__(self,a):
        self._a = a
    def show(self):
        print("protected variable:",self._a)
class B(A):
    def __init__(self, a):
        super().__init__(a)

    def showB(self):
        print("accessing variable to B class:",self._a)

obj = B(30)
obj.show()
obj.showB()
```
