```python
class RBI: # abstract class
    def interest(self): # abstract method
        pass # abstract method do not contain their implimentation
class SBI(RBI):
    def interest(self):
        print("SBI interest rate is 5 %")
class HDFC(RBI):
    def interest(self):
        print("HDFC interest rate is 7 %")

s = SBI()
s.interest()
h = HDFC()
h.interest()
```

```python
# absract method 
from abc import ABC, abstractmethod
class Operation(ABC):
    @abstractmethod
    def calculate(self,a,b):
        pass
class Addition(Operation):
    def calculate(self,a,b):
        return f"addition {a+b}"
class Subtraction(Operation):
    def calculate(self,a,b):
        return f"Subtraction {a-b}"
class multiplication(Operation):
    def calculate(self,a,b):
        return f"multiplication {a*b}"
class devision(Operation):
    def calculate(self,a,b):
        return f"devision {a/b}"
add = Addition()
sub = Subtraction()
mult =multiplication()
div = devision()
a = int(input("give the value of a: "))
b = int(input("give the value of b: "))

print(add.calculate(a,b))
print(sub.calculate(a,b))
print(mult.calculate(a,b))
print(div.calculate(a,b))
```

```python
# ATM machine
from abc import ABC, abstractmethod
class Bank(ABC):
    @abstractmethod
    def check_balance(self):
        pass
    @abstractmethod
    def withdraw_money(self,amount):
        pass
class BankATM(Bank):
    def __init__(self,balance):
        self.balance = balance
        
    def check_balance(self):
        return f"your balance is : {self.balance}"
    
    def withdraw_money(self,amount):
        if amount > self.balance:
            return "insufficient balance"
        self.balance -=amount
        return f"withdrawl successfull! new balance{self.balance}"

initial_bal = float(input("enter the initial balance: "))
atm = BankATM(initial_bal)
print(atm.check_balance())
print(atm.withdraw_money(500))
```
