```python
# how to create object?
```

```python
class myclass:
    x = 5 # property of class
    y = 7
# create object for myclass
m1 = myclass() # m1 is an object of myclass
print(m1.x)
print(m1.y)
```

```python
# calling a function by object
class myclass:
    def myfunction(self): # property of myclass
        print("hello my function")
# creating an object
m1 = myclass()
m1.myfunction()
```

```python
class mobile:
    def camera(self):
        print("camera is opening")
    def mice(self):
        print("i am listening your word")
    def instagram(self):
        print("i am showing you reeels")
y = mobile() # y is a object created for mobile class
y.mice()
```

```python
# constructor and function
class myclass:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def myFunction(self):
        print("my name is: ",self.name)
# object
my = myclass("yogesh",26)
print(my.age)
print(my.name)
my.myFunction()
# changing the age using object
my.age = 27
print(my.age)
# change name
my.name = "person name"
print(my.name)
# delete the object
del my.age
print("age is deleted so i will not show the age")
print(my.name)
# print(my.age)
```
