```python
# read() method reads the entire file and returns it as a string
# r mode is for reading
f = open("file.txt","r")
content = f.read()
print(content)
f.close()
```

```python
# write() method it will replace the existing content of the file with the new content. If the file does not exist, it will create a new file.
# w mode is for writing
f = open("writefile.txt","w")
f.write("This is my programming file")
f.close()
```

```python
content = "hey this is yogesh a python programmer"
f = open("writefile.txt","w")
f.write(content)
f.close()
```

```python
# append() method it will add the new content to the existing content of the file. If the file does not exist, it will create a new file.
# a mode is for appending
add = "This is my programming file"
f = open("writefile.txt","a")
f.write(add + "\n")
f.close()
```

```python
# readline() method reads a single line from the file. If you call it again, it will read the next line.
# readlines() method reads all the lines of a file and returns them as a list of strings.
```

```python
# readlines()
f = open("writefile.txt")
lines = f.readlines()
print(lines)
print(type(lines))
```

```python
# readline()
f = open("writefile.txt")
line1 = f.readline()
print(line1)
line2 = f.readline()
print(line2)
line3 = f.readline()
print(line3)
print(type(line3))
line4 = f.readline()
print(line4)
line5 = f.readline()
print(line5)
```

```python
# readline() with while loop
f = open("writefile.txt")
line = f.readline()
while(line != ""):
    print(line)
    line = f.readline()
f.close()
```

```python
# with statement is used to open a file and automatically close it after the block of code is executed. It is a good practice to use with statement when working with files.
with open("writefile.txt") as any_variable_name:
    print(any_variable_name.read())
    
```
