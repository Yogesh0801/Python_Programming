```python
# A list contains the multiplication table of 7. Write a program to convert it to vertical string
# of same numbers.
# 7
# 14
# .
# .
# .
table = [str(7*i) for i in range(1,11)]
print(table)

j = "\n".join(table)
print(j)
```

```python
# Write a program to filter a list of numbers which are divisible by 5.
def divisible5(n):
    if n%5==0:
        return True
    else:
        return False
a = [1,2,344332,545454,555,57565,65,5]
f = list(filter(divisible5,a))
print(f)
```

```python
# Write a program to find the maximum of the numbers in a list using the reduce function.

from functools import reduce # reduce(function, iterable, initializer)
l = [111,2,65,635,74,88,222,312]
def greaterNO(x,y):
    if x>y:
        return x
    return y
print(reduce(greaterNO,l))
```

```python
# 8 .Find students who scored more than 60:
marks = [50, 60 , 70 , 80 ,90]
result = lambda x : x > 60
f = filter(result, marks)
print(list(f))
```

```python
# 9 Calculate total marks:
# 	50 + 60 + 70 + 80 + 90 = 350
# reduce
from functools import reduce
marks = [50, 60 , 70 , 80 ,90]
total = lambda x, y : x + y
total_marks = reduce(total, marks)
print(total_marks)
```
