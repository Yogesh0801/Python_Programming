```python
# try and except block to handle the exception
try:
    print(10/0)
except ZeroDivisionError:
    print("You cannot divide by zero!")
```

```python
n = int(input("Enter a number: "))
print(n)
try:
    print(10/n)
except ZeroDivisionError:
    print("You cannot divide by zero!")
```

```python
try:
    n = int(input("enter a number: "))
    m = int(input("enter other number: "))
    print(n/m)
except ZeroDivisionError:
    print("cant devide by zero....")
except ValueError:
    print("please provide int value only")
```

```python
try:
    n = int(input("enter a number: "))
    m = int(input("enter other number: "))
    print(n/m)
except ArithmeticError:
    print("arithmatic error")
except ValueError:
    print("please provide int value only")
```

```python
# finally block : this block will run compulsary if the error accurs no not
x = 5657
try:
    print(x)
except:
    print("something went wrong")
finally:
    print("finally block is working")
```
