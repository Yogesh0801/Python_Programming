```python
# What will be the length of following set s:
s = set()
s.add(20)
s.add(20.0)
s.add('20') # length of s after these operations?
print(s)
print(len(s))#20, 20.0 are same so length will be 2
```

```python
# Create an empty dictionary. Allow 4 friends to enter their favorite language as value and
# use key as their names. Assume that the names are unique.
d = {}
name1 = input("enter name of friend 1: ")
lang1 = input("enter favorite language of friend 1: ")
d.update({name1:lang1}) 

name2 = input("enter name of friend 2: ")
lang2 = input("enter favorite language of friend 2: ")
d.update({name2:lang2}) 

name3 = input("enter name of friend 3: ")
lang3 = input("enter favorite language of friend 3: ")
d.update({name3:lang3}) 

name4 = input("enter name of friend 4: ")
lang4 = input("enter favorite language of friend 4: ")
d.update({name4:lang4}) 

print(d)
```
