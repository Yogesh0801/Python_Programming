```python
# parameterized function
def wish(name):
    print("hello",name,"good morning")
wish("yogesh")
wish("subha")
wish("annu")
wish("maithli")
```

```python
def wish(name, age):
    print("hello",name,"good morning, your age is",age)

wish("yogesh", 25)
wish("subha", 24)
wish("annu", 23)
wish("maithli", 22)
```

```python
# write a function for factorial of a number
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
for i in range(1, 6):
    print(f"The factorial of {i} is {factorial(i)}")
print(factorial(5))  
```

```python
def factorial(n):
    result = 1 #5,20,60,120,120
    while n >=1:
        result *= n #5,4,3,2,1
        n -= 1 #4,3,2,1,0
    return result
print(factorial(5))
```

```python
# recursion: a function that calls itself is known as recursion. It is used to solve problems that can be broken down into smaller, similar subproblems. Recursion is often used in algorithms and data structures, such as tree traversal and sorting algorithms.
# factorial of 5 : 5! = 5 × 4 × 3 × 2 × 1 = 120
# factorial(n) = n * factorial(n-1)
#               = 5*4*3*2*1
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1) # recursive call # 20,60,120
n = int(input("Enter a number: "))
print(factorial(n))
```
