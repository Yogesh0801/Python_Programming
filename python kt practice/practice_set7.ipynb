```python
# shyam is reading a mobile number as words and he want to copy the number in digits. Write a program to convert the words into digits.
# ex: "nine four three" -> "943"
```

```python
words = input("enter your mobile number in word: ").lower()
word_list = words.split()
print(word_list)
num = ""
for word in word_list:
    if word == "one":
        num += "1"
    elif word == "two":
        num += "2"
    elif word == "three":
        num += "3"
    elif word == "four":
        num += "4"
    elif word == "five":
        num += "5"
    elif word == "six":
        num += "6"
    elif word == "seven":
        num += "7"
    elif word == "eight":
        num += "8"
    elif word == "nine":
        num += "9"
    elif word == "zero":
        num += "0"
    else:
        print("you have enter wrong number")
print("numeric mobile number is: ",num)
```

```python
# Write a program to find whether a given number is prime or not.
# prime number is a number which is only divisible by 1 and itself.
# 11/1 11/11
# 9/1 9/3 9/9 not prime
n = int(input("enter a number: "))
print(n)
for i in range(2,n):
    print(n,"%",i,"is",n%i)
    if (n % i)==0:
        print(n,"is not a prime number")
        break
else:
    print(n,"is a prime number")

```

```python
# Write a program to find the sum of first n natural numbers using while loop.
n = int(input("enter a number: "))
print("the value you are given is: ", n)
i = 1
sum = 0
while i <= n:
    print("i value ", i)
    sum += i
    print("sum is: ", sum)
    i += 1
```

```python
# Write a program to calculate the factorial of a given number using for loop.
# 5 = 5*4*3*2*1
n = int(input("enter a number: "))
print("the value given by user is: ", n)
fact = 1
for i in range(1,n+1):
    fact *= i

    print(f"i th value {i} fact value {fact} ")
print("the factorial of ", n, "is: ", fact)
```
