```python
# The game() function in a program lets a user play a game and returns the score as an
# integer. You need to read a file ‘Hi-score.txt’ which is either blank or contains the previous
# Hi-score. You need to write a program to update the Hi-score whenever the game()
# function breaks the Hi-score.
import random # random module will generate the random number
# print(random.randint(1,100))
def game():
    print("Welcome to the game!")
    score = random.randint(1, 100)
    with open('Hi-score.txt') as f:
        hi_score = f.read()
        if hi_score != '':
            hi_score = int(hi_score)
        else:
            hi_score = 0
    print(f"Your score is: {score}")
    if score > hi_score:
        print("Congratulations! You have broken the Hi-score!")
        with open('Hi-score.txt', 'w') as f:
            f.write(str(score))
    else:
        print("you lost! Better luck next time.")
        
        print(f"The Hi-score is: {hi_score}")
    return score
game()
```

```python
# Write a program to generate multiplication tables from 2 to 20 and write it to the different
# files. Place these files in a folder table.
def generateTable(n):
    table = ""
    for i in range(1,11):
        table += f"{n} x {i} = {n*i}\n"
    with open(f"tables/table_{n}.txt",'w') as f:
        f.write(table)
for i in range(2,21):
    generateTable(i)
```

```python
# A file contains a word “Donkey” multiple times. You need to write a program which
# replaces this word with ##### by updating the same file.
word = "Donkey"
with open('file.txt') as f:
    content = f.read()
new_content = content.replace(word, "#####")
with open('file.txt', 'w') as f:
    f.write(new_content)

print(content)
```

```python
# Repeat program 4 for a list of such words to be censored.
words = ["Donkey", "stupid", "idiot"]
with open('file.txt') as f:
    content = f.read()

for word in words:
    content = content.replace(word, "#####")

with open('file.txt', 'w') as f:
    f.write(content)
print(content)
```
