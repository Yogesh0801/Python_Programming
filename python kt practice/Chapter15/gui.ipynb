```python
# make a gui for my calculator
from tkinter import *
def add():
    num1 = int(number1.get())
    num2 = int(number2.get())
    result = num1 + num2
    outputText.delete(1.0, END)
    outputText.insert(END, f"Addition: {result}")

def sub():
    num1 = int(number1.get())
    num2 = int(number2.get())
    result = num1 - num2
    outputText.delete(1.0, END)
    outputText.insert(END, f"\nSubtraction: {result}")

def multiplication():
    num1 = int(number1.get())
    num2 = int(number2.get())
    result = num1 * num2
    outputText.delete(1.0, END)
    outputText.insert(END, f"\nMultiplication: {result}")

def divide():
    num1 = int(number1.get())
    num2 = int(number2.get())
    try:
        result = num1 / num2
        outputText.insert(END, f"\nDivision: {result}")
    except ZeroDivisionError:
        outputText.delete(1.0, END)
        outputText.insert(END, f"\nDivide by zero")

def modulus():
    num1 = int(number1.get())
    num2 = int(number2.get())
    try:
        result = num1 % num2
        outputText.delete(1.0, END)
        outputText.insert(END, f"\nModulus: {result}")
    except ZeroDivisionError:
        outputText.delete(1.0, END)
        outputText.insert(END, f"\nDivide by zero")

def square():
    num1 = int(number1.get())
    result = num1 ** 2
    outputText.delete(1.0, END)
    outputText.insert(END, f"\nSquare: {result}")
   
r = Tk()
r.title("calculator")
r.geometry("800x1000")
r.config(bg="orange")
number1 = StringVar()
number2 = StringVar()
l1 = Label(r, text="CALCULATOR", font=("times", 30, "bold"), fg="black")
l1.place(x=250, y=30)
l2 = Label(r, text="Enter Number 1", font=("chiller", 20, "bold"), fg="black")
l2.place(x=150, y=100)
l3 = Label(r, text="Enter Number 2", font=("chiller", 20, "bold"), fg="black")
l3.place(x=150, y=150)
e1 = Entry(r, textvariable=number1, font=("chiller", 20, "bold"), fg="black")
e1.place(x=400, y=100)
e2 = Entry(r,textvariable=number2, font=("chiller", 20, "bold"), fg="black")
e2.place(x=400, y=150)
outputText = Text(r, height=5,width=25, font=("elephant", 20, "bold"), fg="black")
outputText.place(x=150, y=230)
b1 = Button(r, text="+", command = add,font=("elephant", 20, "bold"), fg="black").place(x=150, y=450)
b2 = Button(r, text="-", command = sub,font=("elephant", 20, "bold"), fg="black").place(x=230, y=450)
b3 = Button(r, text="*", command = multiplication,font=("elephant", 20, "bold"), fg="black").place(x=280, y=450)
b4 = Button(r, text="/", command = divide,font=("elephant", 20, "bold"), fg="black").place(x=350, y=450)
b5 = Button(r, text="%", command = modulus,font=("elephant", 20, "bold"), fg="black").place(x=400, y=450)
b6 = Button(r, text="**", command = square,font=("elephant", 20, "bold"), fg="black").place(x=470, y=450)
b6 = Button(r, text="Exit",command=r.destroy, font=("chiller", 20, "bold"),
             fg="black").place(x=550, y=450)

mainloop()
```
