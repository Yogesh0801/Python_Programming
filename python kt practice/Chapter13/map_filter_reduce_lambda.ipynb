```python
# map()
# apply the operation to every element
# without map
l = [1,2,3,4,5]
sqlist = []
for i in l:
    sqlist.append(i*i)
    # print(i)
print(sqlist)
```

```python
# using map
l = [1,2,3,4,5]
def square(x):
    return x*x
sqlist = map(square, l)
print(list(sqlist))
```

```python
# without lambda
def square(x):
    return x*x
square(4)
```

```python
# with lambda
square = lambda x : x*x
print(square(5))
```

```python
add = lambda x,y : x + y
print(add(2, 3))
```

```python
l = [1, 2, 3, 4, 5]
square = lambda x : x * x
sqlist = map(square, l)
print(list(sqlist))
```

```python
# filter()
# filter the elements of a list
# select the elements that satisfy a condition
```

```python
# without filter
l = [1, 2, 3, 4, 5]
even_number = []
odd_number = []
for i in l:
    if i % 2 == 0:
        even_number.append(i)
    else:
        odd_number.append(i)
    # print(i)
print(f"even_number{even_number}") 
print(f"odd_number{odd_number}")
```

```python
# using filter() : syntax-filter(function, list)
l = [1, 2, 3, 4, 5]
def even(x):
    if x % 2 == 0:
        return True
    else:
        return False
evenlist = filter(even,l)
print(list(evenlist))
```

```python
# using lambda function
l = [1, 2, 3, 4, 5]
evenlist = filter(lambda x : x % 2 == 0, l)
print(list(evenlist))
```

```python
# reduce function : reduce(function, iterable)
# take multiple elements from iterable and apply function to them and produce a single element
```

```python
# without reduce
l = [1, 2, 3, 4, 5]
total = 0
for i in l:
    # total += i
    total = total + i
    print(i)
print(total)
```

```python
# with reduce() syntax : reduce(function, iterable)
from functools import reduce
l = [1, 2, 3, 4, 5]
def add(x,y):
    return x + y
print(reduce(add, l))
```

```python
# with lambda
l = [1, 2, 3, 4, 5]
add  = lambda x, y: x + y
print(reduce(add, l))
```

```python
l = [1, 2, 3, 4, 5]
def multiply(x, y):
    return x * y
print(reduce(multiply, l))
```

```python
from functools import reduce
l = [1, 2, 3, 4, 5]
multiply = lambda x, y: x * y
print(reduce(multiply, l))
```
