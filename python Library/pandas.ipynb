```python
# replace the value based on dataset using Mean, Median and Mode
# mean : sum of all values divided by the number of values
# median : middle value of the sorted values in ascending order
# mode : most frequent value

import pandas as pd
data = [2,2,3,4,5,7,7]
df = pd.DataFrame(data)

print(sum(data)/len(data))
print(df.mean())
```

```python
print(df.median())
```

```python
print(df.mode())
```

```python
df = pd.read_csv('data (1).csv')
print(df)
```

```python
# fill max pulse using median
# fill Duration using mode
```

```python
# gigo(garbage in garbage out)
med = df['Maxpulse'].median()
print(med)
df.fillna({'Maxpulse':med},inplace=True)
print(df)
```

```python
md = df['Duration'].mode()[0] # mode of duration
print(md)
df.fillna({'Duration':md}, inplace=True)
print(df)
```

```python
# duplicate
print(df.duplicated())
```

```python
df.drop_duplicates()
```

```python
# correlation : corr()
df = pd.read_csv('data.csv')
df
```

```python
df.corr()
# the number varies from 0 to 1
# 1 means perfect correlation
# 0 means no correlation
# negative correlation means inverse correlation
# 0.9 means 90% correlation
# 0.2 means 20% correlation
# good correlation:  is 0.6 (-0.6) and above 
```
