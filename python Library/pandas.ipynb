```python
# Series: it is a one-dimensional labeled array capable of holding any data type.
```

```python
import pandas as pd
a = [1,2,"h",4.6]
# print(a)
ds = pd.Series(a)
print(ds)
```

```python
print(ds[2])
```

```python
# Labels:
# if nothing is specified, the default is to use the index 0, 1, 2, etc.
```

```python
import pandas as pd
a = [1,2,"h",4.6]
ds = pd.Series(a,index=['a','b','c','d'])
print(ds)
```

```python
print(ds['c'])
```

```python
# using the key/value create as series
calaries = {
    "day1":420,
    "day2":380,
    "day3":390}
sd = pd.Series(calaries)
print(sd)
```

```python
data = {
    "calories" :[420,380,390],
    "duration":[50,40,34]
}
ds = pd.Series(data)
print(ds)
df = pd.DataFrame(data)
print(df)
```

```python
print(ds['calories'])
```
